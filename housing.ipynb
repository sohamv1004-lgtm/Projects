```python
def is_convert(x):
    if isinstance(x, (int, float)):
        return x

    if isinstance(x, str):
        if '-' in x:
            a, b = x.split('-')
            return (float(a) + float(b)) / 2
        try:
            return float(x)
        except:
            return None
```

```python
is_convert('sq')
```

```python
new_df['total_sqft']=new_df['total_sqft'].apply(is_convert)
new_df
```

```python
df1=new_df.copy()
df1['price_per_sqft']=df1['price']*100000/df1['total_sqft']
df1
```

```python
df1['location']=df1['location'].apply(lambda x: x.strip())
location_stat=df1.groupby('location')['location'].agg('count')<10
```

```python
location_10 = location_stat[location_stat].index
df1['location'] = df1['location'].apply(
    lambda x: 'other' if x in location_10 else x
)
df1['location'].value_counts()
```

```python
df2 = df1[~((df1['total_sqft'] / df1['bhk']) <300)]
df2.shape
```

```python
df2['price_per_sqft'].describe()
```

```python
def remove_pps_outliers(df):
    df_out = pd.DataFrame()
    for key, subdf in df.groupby('location'):
        m = np.mean(subdf.price_per_sqft)
        st = np.std(subdf.price_per_sqft)
        reduced_df = subdf[(subdf.price_per_sqft>(m-st)) & (subdf.price_per_sqft<=(m+st))]
        df_out = pd.concat([df_out,reduced_df],ignore_index=True)
    return df_out
df3 = remove_pps_outliers(df2)
df3
```

```python
def remove_bhk_outliers(df):
    exclude_indices = np.array([])
    for location, location_df in df.groupby('location'):
        bhk_stats = {}
        for bhk, bhk_df in location_df.groupby('bhk'):
            bhk_stats[bhk] = {
                'mean': np.mean(bhk_df.price_per_sqft),
                'std': np.std(bhk_df.price_per_sqft),
                'count': bhk_df.shape[0]
            }
        for bhk, bhk_df in location_df.groupby('bhk'):
            stats = bhk_stats.get(bhk-1)
            if stats and stats['count']>5:
                exclude_indices = np.append(exclude_indices, bhk_df[bhk_df.price_per_sqft<(stats['mean'])].index.values)
    return df.drop(exclude_indices,axis='index')
df4 = remove_bhk_outliers(df3)

df4.shape
```

```python
f_df=df4.drop(['size','price_per_sqft'],axis='columns')
f_df
```

```python
dummies = pd.get_dummies(f_df.location)
dummies.head(3)
```

```python
df6 = pd.concat([f_df,dummies.drop('other',axis='columns')],axis='columns')
df6.head()

```

```python
df7=df6.drop(['location'],axis='columns')
df
```

```python
X = df7.drop(['price'],axis='columns')
X.head(3)
```

```python
y = df7.price
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=10)
```

```python
from sklearn.linear_model import LinearRegression
lr_clf = LinearRegression()
lr_clf.fit(X_train,y_train)
lr_clf.score(X_test,y_test)  

```

```python
def predict_price(location,sqft,bath,bhk):    
    loc_index = np.where(X.columns==location)[0][0]

    x = np.zeros(len(X.columns))
    x[0] = sqft
    x[1] = bath
    x[2] = bhk
    if loc_index >= 0:
        x[loc_index] = 1

    return lr_clf.predict([x])[0]
```
