```python
total_score_df1=df1.groupby(['match_id','inning']).sum()['total_runs'].reset_index()
```

```python

total_score_df1 = total_score_df1[total_score_df1['inning'] == 1]
total_score_df1
```

```python
match_df = df.merge(total_score_df1[['match_id','total_runs']],left_on='id',right_on='match_id')
```

```python
match_df
```

```python
teams = [
    'Sunrisers Hyderabad',
    'Mumbai Indians',
    'Royal Challengers Bangalore',
    'Kolkata Knight Riders',
    'Kings XI Punjab',
    'Chennai Super Kings',
    'Rajasthan Royals',
    'Delhi Capitals'
]
```

```python
match_df['team1'] = match_df['team1'].str.replace('Deccan Chargers','Sunrisers Hyderabad')
match_df['team2'] = match_df['team2'].str.replace('Deccan Chargers','Sunrisers Hyderabad')
```

```python
match_df = match_df[match_df['team1'].isin(teams)]
match_df = match_df[match_df['team2'].isin(teams)]
```

```python
match_df.shape
```

```python
new_match_df=match_df[match_df['dl_applied']==0]
new_match_df
```

```python
new_match_df=new_match_df[['match_id','total_runs','winner','city']]
```

```python
delivery_df = new_match_df.merge(df1,on='match_id')
delivery_df
```

```python
delivery_df = delivery_df[delivery_df['inning'] == 2]
```

```python
delivery_df
```

```python
delivery_df['current_score'] = (
    delivery_df.groupby('match_id')['total_runs_y'].cumsum()
)
```

```python
delivery_df['runs_left']=delivery_df['total_runs_x']-delivery_df['current_score']
delivery_df
```

```python
delivery_df['ball'].unique()
```

```python
delivery_df['balls_left']=126-(delivery_df['over']*6+delivery_df['ball'])
delivery_df
```

```python
delivery_df['player_dismissed'] = delivery_df['player_dismissed'].fillna("0")
delivery_df['player_dismissed'] = delivery_df['player_dismissed'].apply(lambda x:x if x == "0" else "1")
delivery_df['player_dismissed'] = delivery_df['player_dismissed'].astype('int')
```

```python
delivery_df
```

```python
delivery_df.dtypes
```

```python
delivery_df['wickets_fallen'] = (
    delivery_df.groupby(['match_id', 'inning'])['player_dismissed']
    .cumsum()
)

delivery_df['wickets'] = 10 - delivery_df['wickets_fallen']
delivery_df
```

```python
delivery_df['crr'] = (
    delivery_df['current_score'] / (120 - delivery_df['balls_left'])
) * 6
```

```python
delivery_df
```

```python
delivery_df['rrr']=(delivery_df['runs_left']*6/delivery_df['balls_left'])
delivery_df
```

```python
def result(row):
    return 1 if row['batting_team'] == row['winner'] else 0
```

```python
delivery_df['result'] = delivery_df.apply(result,axis=1)
```

```python
delivery_df
```

```python
final_df = delivery_df[['batting_team','bowling_team','city','runs_left','balls_left','wickets','total_runs_x','crr','rrr','result']]
final_df = final_df.sample(final_df.shape[0])
final_df.sample()
```

```python
final_df.dropna(inplace=True)
final_df = final_df[final_df['balls_left'] != 0]
final_df
```

```python
X = final_df.drop('result', axis=1)
y = final_df['result']

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=1)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

trf = ColumnTransformer([
    ('trf',OneHotEncoder(sparse_output=False,drop='first'),['batting_team','bowling_team','city'])
]
,remainder='passthrough')
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
```

```python
pipe = Pipeline(steps=[
    ('step1',trf),
    ('step2',RandomForestClassifier())
])
```

```python
pipe.fit(X_train,y_train)
```

```python
y_pred = pipe.predict(X_test)
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
